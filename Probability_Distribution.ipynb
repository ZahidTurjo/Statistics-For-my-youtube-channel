### Poisson Distribution

 Suppose you are a Network Administrator for a small company. You’ve tracked your data and found that, on average, your server experiences 2 minor glitches per day ($\lambda = 2$).You want to know the probability that on a particularly unlucky day, the server will have exactly 4 glitches ($k = 4$).

```python
from scipy.stats import poisson

lam = 2
k = 4

probability = poisson.pmf(k, lam)

print("Probability of exactly 4 glitches:", probability)
```

```python
probs=[]
for i in range(11):
  pro=poisson.pmf(i,lam)
  probs.append(pro)
```

```python
import pandas as pd
pd.Series(probs).plot(kind='bar')
```

# Binomial Distribution

```python
import numpy as np
n = 6        # number of trials
p = 0.8      # probability of success
experiment = 1000000

data = np.random.binomial(1, p, size=(experiment, n))
print(data)
```

```python
data.shape
```

```python
data.sum(axis=1)
```

```python
# Count number of good bulbs in each experiment
success_counts=data.sum(axis=1)
print(success_counts)
len(success_counts)
```

```python
# Count how many times exactly 4 successes occur
count_4 = np.sum(success_counts == 4)
print(count_4)
```

image.png

```python
probability = count_4 / experiment

print("Estimated Probability:", probability)
```

```python
import matplotlib.pyplot as plt

plt.hist(success_counts, bins=range(8), align='left')
plt.xlabel("Number of Good Bulbs")
plt.ylabel("Frequency")
plt.title("Binomial Simulation (n=6, p=0.8)")
plt.show()
```
